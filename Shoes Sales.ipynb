# “Comparative Analysis of Revenue Across Sales Channels”

The analysis below explores how revenue varies across different sales channels Mall, Online, and Retail Store. By grouping the data based on the sales channel and calculating the average revenue, a clear pattern emerges.
The Retail Store channel stands out as the top performer, generating the highest average revenue of approximately 1493 USD. This suggests that customers in physical retail environments tend to spend more, possibly due to in-store experiences, promotions, or product visibility.
The Mall channel follows with an average revenue of about 1452 USD, indicating stable performance and consistent customer spending. Malls continue to play an important role in driving sales, although not as strongly as standalone retail stores.
In contrast, the Online channel records the lowest average revenue at around 1413 USD. This may reflect lower transaction values online, possibly due to price comparisons, discounts, or less impulse buying compared to physical locations.
The point plot visually confirms this trend, showing a noticeable dip for Online sales and a peak for Retail Store sales.

```python
df2= df.groupby('Sales_Channel')['Revenue_USD'].mean()
df2
```

```python
sns.pointplot(df2)
```

# “Daily Revenue Trends Analysis Over Time”

The analysis examines how total revenue changes over time across the year by aggregating daily sales. By grouping revenue by date and visualizing it using a line plot, a dynamic picture of business performance over time is revealed.
At the beginning of the year, revenue shows moderate activity, with some days performing better than others. As time progresses, the pattern becomes more volatile, with sharp peaks and dips throughout the year. These fluctuations indicate that sales are not consistent on a daily basis and are influenced by varying factors such as demand, promotions, or seasonal effects.
Notably, there are several significant spikes, where revenue reaches unusually high levels (close to 20,000 USD at some points). These peaks may correspond to special events, promotional campaigns, or high-demand periods such as holidays or sales seasons.
On the other hand, there are also low-revenue days, where sales drop significantly, suggesting periods of weak demand or limited business activity.
Overall, the trend does not follow a smooth upward or downward direction but instead shows high variability, indicating that the business experiences frequent changes in daily performance rather than steady growth.

# Pair Plot Summary (Price, Units Sold, and Revenue)

The pair plot provides a comprehensive view of the relationships between Price_USD, Units_Sold, and Revenue_USD, highlighting both distributions and correlations.

The diagonal histograms show that Price_USD is fairly spread out across a wide range, suggesting products are sold at different price levels. Units_Sold appears evenly distributed, indicating that sales volumes are relatively consistent across transactions. In contrast, Revenue_USD is right-skewed, meaning most values are lower, with fewer high-revenue observations.

Looking at the scatter plots, there is a strong positive relationship between Units_Sold and Revenue_USD. As the number of units sold increases, revenue also increases significantly, which is expected since revenue is directly influenced by sales volume.

There is also a positive relationship between Price_USD and Revenue_USD, though it is more spread out. Higher-priced items can generate higher revenue, but the effect depends on how many units are sold.
However, the relationship between Price_USD and Units_Sold appears weaker, suggesting that higher prices do not always lead to fewer sales, and customer demand is not strictly dependent on price alone.

Key Insight:
Revenue is primarily driven by units sold, with price playing a secondary role. The combined effect of both variables determines overall sales performance, with high revenue achieved when both price and quantity are high.
